# Linear Regression with Noise and Data Export
This notebook demonstrates linear, non-linear, and circular graphs with added random noise, and exports data to CSV.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)
```

## 1. Linear Graph with Noise: y = 2x + 3

```python
# Generate x values
x_linear = np.linspace(-10, 10, 100)
# Generate y values with linear relationship
y_linear_clean = 2*x_linear + 3
# Add random noise
noise_linear = np.random.normal(0, 3, len(x_linear))  # Gaussian noise with std=3
y_linear_noisy = y_linear_clean + noise_linear

# Plot both clean and noisy versions
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(x_linear, y_linear_clean, label='Clean Data', alpha=0.6, s=20)
plt.plot(x_linear, y_linear_clean, 'r-', linewidth=2, label='True Line')
plt.title("Linear Graph: y = 2x + 3 (Clean)")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.scatter(x_linear, y_linear_noisy, label='Noisy Data', alpha=0.6, s=20, color='orange')
plt.plot(x_linear, y_linear_clean, 'r-', linewidth=2, label='True Line')
plt.title("Linear Graph: y = 2x + 3 (With Noise)")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# Create DataFrame for linear data
df_linear = pd.DataFrame({
    'x': x_linear,
    'y_clean': y_linear_clean,
    'noise': noise_linear,
    'y_noisy': y_linear_noisy
})

print("Linear Data (first 10 rows):")
print(df_linear.head(10))
```

## 2. Non-Linear Graph with Noise: y = x²

```python
# Generate x values
x_nonlinear = np.linspace(-10, 10, 100)
# Generate y values with quadratic relationship
y_nonlinear_clean = x_nonlinear**2
# Add random noise (proportional to y values for realistic noise)
noise_nonlinear = np.random.normal(0, 5, len(x_nonlinear))  # Gaussian noise
y_nonlinear_noisy = y_nonlinear_clean + noise_nonlinear

# Plot both clean and noisy versions
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(x_nonlinear, y_nonlinear_clean, label='Clean Data', alpha=0.6, s=20)
plt.plot(x_nonlinear, y_nonlinear_clean, 'g-', linewidth=2, label='True Curve')
plt.title("Non-Linear Graph: y = x² (Clean)")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.scatter(x_nonlinear, y_nonlinear_noisy, label='Noisy Data', alpha=0.6, s=20, color='purple')
plt.plot(x_nonlinear, y_nonlinear_clean, 'g-', linewidth=2, label='True Curve')
plt.title("Non-Linear Graph: y = x² (With Noise)")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# Create DataFrame for non-linear data
df_nonlinear = pd.DataFrame({
    'x': x_nonlinear,
    'y_clean': y_nonlinear_clean,
    'noise': noise_nonlinear,
    'y_noisy': y_nonlinear_noisy
})

print("Non-Linear Data (first 10 rows):")
print(df_nonlinear.head(10))
```

## 3. Circular Graph with Noise

```python
# Generate circular data
r = 5
theta = np.linspace(0, 2*np.pi, 200)

x_circle_clean = r * np.cos(theta)
y_circle_clean = r * np.sin(theta)

# Add random noise to circular coordinates
noise_x = np.random.normal(0, 0.5, len(theta))
noise_y = np.random.normal(0, 0.5, len(theta))

x_circle_noisy = x_circle_clean + noise_x
y_circle_noisy = y_circle_clean + noise_y

# Plot both clean and noisy versions
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(x_circle_clean, y_circle_clean, 'b-', linewidth=2, label='Clean Circle')
plt.scatter(x_circle_clean, y_circle_clean, s=20, alpha=0.6)
plt.title("Circular Graph (Clean)")
plt.xlabel("X")
plt.ylabel("Y")
plt.axis("equal")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(x_circle_clean, y_circle_clean, 'b--', linewidth=1, alpha=0.5, label='True Circle')
plt.scatter(x_circle_noisy, y_circle_noisy, s=20, alpha=0.6, color='cyan', label='Noisy Data')
plt.title("Circular Graph (With Noise)")
plt.xlabel("X")
plt.ylabel("Y")
plt.axis("equal")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# Create DataFrame for circular data
df_circular = pd.DataFrame({
    'theta': theta,
    'x_clean': x_circle_clean,
    'y_clean': y_circle_clean,
    'noise_x': noise_x,
    'noise_y': noise_y,
    'x_noisy': x_circle_noisy,
    'y_noisy': y_circle_noisy
})

print("Circular Data (first 10 rows):")
print(df_circular.head(10))
```

## 4. Export Data to CSV Files

```python
# Export each dataset to CSV
df_linear.to_csv('linear_data.csv', index=False)
df_nonlinear.to_csv('nonlinear_data.csv', index=False)
df_circular.to_csv('circular_data.csv', index=False)

print("✓ CSV files created successfully!")
print(f"  - linear_data.csv ({len(df_linear)} rows)")
print(f"  - nonlinear_data.csv ({len(df_nonlinear)} rows)")
print(f"  - circular_data.csv ({len(df_circular)} rows)")

# Create a combined CSV with all data labeled
df_linear['type'] = 'linear'
df_nonlinear['type'] = 'nonlinear'
# Prepare circular data for combining
df_circular_combined = df_circular[['x_noisy', 'y_noisy']].copy()
df_circular_combined.columns = ['x', 'y_noisy']
df_circular_combined['y_clean'] = df_circular['y_clean']
df_circular_combined['noise'] = np.sqrt(df_circular['noise_x']**2 + df_circular['noise_y']**2)
df_circular_combined['type'] = 'circular'

# Reorder columns for consistency
df_linear_combined = df_linear[['x', 'y_clean', 'noise', 'y_noisy', 'type']]
df_nonlinear_combined = df_nonlinear[['x', 'y_clean', 'noise', 'y_noisy', 'type']]
df_circular_combined = df_circular_combined[['x', 'y_clean', 'noise', 'y_noisy', 'type']]

df_combined = pd.concat([df_linear_combined, df_nonlinear_combined, df_circular_combined], ignore_index=True)
df_combined.to_csv('all_data_combined.csv', index=False)

print(f"  - all_data_combined.csv ({len(df_combined)} rows)")
print("\nCombined data summary:")
print(df_combined.groupby('type').size())
```

## 5. Summary Statistics

```python
print("=" * 60)
print("SUMMARY STATISTICS")
print("=" * 60)

print("\n1. LINEAR DATA (y = 2x + 3):")
print(f"   Mean noise: {df_linear['noise'].mean():.4f}")
print(f"   Std noise: {df_linear['noise'].std():.4f}")
print(f"   Min y_noisy: {df_linear['y_noisy'].min():.4f}")
print(f"   Max y_noisy: {df_linear['y_noisy'].max():.4f}")

print("\n2. NON-LINEAR DATA (y = x²):")
print(f"   Mean noise: {df_nonlinear['noise'].mean():.4f}")
print(f"   Std noise: {df_nonlinear['noise'].std():.4f}")
print(f"   Min y_noisy: {df_nonlinear['y_noisy'].min():.4f}")
print(f"   Max y_noisy: {df_nonlinear['y_noisy'].max():.4f}")

print("\n3. CIRCULAR DATA (r = 5):")
noise_magnitude = np.sqrt(df_circular['noise_x']**2 + df_circular['noise_y']**2)
print(f"   Mean noise magnitude: {noise_magnitude.mean():.4f}")
print(f"   Std noise magnitude: {noise_magnitude.std():.4f}")
print("\n" + "=" * 60)
```
