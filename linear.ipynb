```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)


x = np.linspace(1, 10, 100)
noise = np.random.normal(0, 3, 100)
y = 3*x + 4 + noise
plt.scatter(x, y,  color="purple", label="data points")

true_y = 3*x + 4
plt.plot(x, true_y, color="red", label="true line")
plt.legend()
plt.title("Linear Graph: y = 3x + 4 + noise")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python


import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

x = np.linspace(-10, 10, 80)
noise = np.random.normal(0, 5, 80)
y = x*x + 2*x + 1 + noise
plt.plot(x, y, 'o')

true_y = x*x + 2*x + 1
plt.plot(x, true_y, color="red", label="true curve")

plt.scatter(x, y)
plt.title("Non-Linear Graph: y = x²")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

r = 5
theta = np.linspace(0, 2*np.pi, 100)
noise_x = np.random.normal(0, 2, 100)
noise_y = np.random.normal(0, 2, 100)
x = r * np.cos(theta) + noise_x
y = r * np.sin(theta) + noise_y

plt.plot(x, y,'o')

true_x = r * np.cos(theta)
true_y = r * np.sin(theta)
plt.plot(true_x, true_y)

plt.scatter(x, y )
plt.title("Circular Graph")
plt.xlabel("X")
plt.ylabel("Y")
plt.axis("equal")
plt.grid(True)
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/newclearbrahma72-svg/Intership/main/IN_AS%20(2).csv"

df = pd.read_csv(url)

df.columns = df.columns.str.strip().str.lower()

# Convert date into datetime format
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Select Assam's data
assam = df[df["subregion1_name"] == "Assam"].copy()

# Remove rows where date or cases are missing
assam = assam.dropna(subset=["date", "new_confirmed"])

# Sort data by date
assam = assam.sort_values("date")

# Calculate a 7-day moving average
assam["average"] = (
    assam["new_confirmed"].rolling(window=7).mean()
)

# Draw the graph
plt.figure(figsize=(12, 5))

plt.plot(
    assam["date"],
    assam["new_confirmed"],
    label="Daily Confirmed Cases",
    alpha=0.5
)

plt.plot(
    assam["date"],
    assam["average"],
    label="7-Day Moving Average",
    linewidth=2
)

plt.xlabel("Date")
plt.ylabel("New Cases")
plt.title("Daily COVID-19 Cases in Assam")

plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
