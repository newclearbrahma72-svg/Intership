```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)


x = np.linspace(1, 10, 100)
noise = np.random.normal(0, 3, 100)
y = 3*x + 4 + noise
plt.scatter(x, y,  color="purple", label="data points")

true_y = 3*x + 4
plt.plot(x, true_y, color="red", label="true line")
plt.legend()
plt.title("Linear Graph: y = 3x + 4 + noise")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python


import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

x = np.linspace(-10, 10, 80)
noise = np.random.normal(0, 5, 80)
y = x*x + 2*x + 1 + noise
plt.plot(x, y, 'o')

true_y = x*x + 2*x + 1
plt.plot(x, true_y, color="red", label="true curve")

plt.scatter(x, y)
plt.title("Non-Linear Graph: y = x²")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

r = 5
theta = np.linspace(0, 2*np.pi, 100)
noise_x = np.random.normal(0, 2, 100)
noise_y = np.random.normal(0, 2, 100)
x = r * np.cos(theta) + noise_x
y = r * np.sin(theta) + noise_y

plt.plot(x, y,'o')

true_x = r * np.cos(theta)
true_y = r * np.sin(theta)
plt.plot(true_x, true_y)

plt.scatter(x, y )
plt.title("Circular Graph")
plt.xlabel("X")
plt.ylabel("Y")
plt.axis("equal")
plt.grid(True)
plt.show()
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


url = "https://raw.githubusercontent.com/newclearbrahma72-svg/Intership/main/IN_AS%20%281%29.csv"


df = pd.read_csv(url)


print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:", df.shape)
print("\nColumn Names:", df.columns.tolist())

```

```python

numeric_data = df.select_dtypes(include=np.number)

print("Numeric Columns:")
print(numeric_data.head())


if numeric_data.shape[1] >= 2:
    x = numeric_data.iloc[:, 0].to_numpy()
    y = numeric_data.iloc[:, 1].to_numpy()

    plt.figure(figsize=(7, 5))
    plt.scatter(x, y)
    plt.xlabel(numeric_data.columns[0])
    plt.ylabel(numeric_data.columns[1])
    plt.title("Real Dataset Visualization")
    plt.grid(True)
    plt.show()
else:
    print("Graph banane ke liye kam se kam 2 numeric columns chahiye.")

```
