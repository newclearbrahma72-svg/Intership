```python

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)


x = np.linspace(1, 10, 100)
noise = np.random.normal(0, 3, 100)
y = 3*x + 4 + noise
plt.scatter(x, y,  color="purple", label="data points")

true_y = 3*x + 4
plt.plot(x, true_y, color="red", label="true line")
plt.legend()
plt.title("Linear Graph: y = 3x + 4 + noise")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

x = np.linspace(-10, 10, 80)
noise = np.random.normal(0, 5, 80)
y = x*x + 2*x + 1 + noise
plt.plot(x, y, 'o')

true_y = x*x + 2*x + 1
plt.plot(x, true_y, color="red", label="true curve")

plt.scatter(x, y)
plt.title("Non-Linear Graph: y = x²")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

r = 5
theta = np.linspace(0, 2*np.pi, 100)
noise = np.random.normal(0, 2, 100)

x = r * np.cos(theta) + noise
y = r * np.sin(theta) + noise
plt.plot(x, y, 'o')

true_x = r * np.cos(theta)
true_y = r * np.sin(theta)
plt.plot(true_x, true_y, color="red", label="true circle")

plt.scatter(x, y )
plt.title("Circular Graph")
plt.xlabel("X")
plt.ylabel("Y")
plt.axis("equal")
plt.grid(True)
plt.show()

```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(0)

x = np.linspace(1, 10, 100)
noise = np.random.normal(0, 3, 100)
y = 3*x + 4 + noise
df = pd.DataFrame({"x": x, "y": y})
df.to_csv("data.csv", index=False)
print(df.head())


print("CSV file create successfully")

```
