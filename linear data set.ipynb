```python

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)


x = np.linspace(1, 10, 100)
noise = np.random.normal(0, 3, 100)
y = 3*x + 4 + noise
plt.plot(x, y, color="purple", label="data points")

true_y = 3*x + 4
plt.plot(x, true_y, color="red", label="true line")
plt.legend()
plt.title("Linear Graph: y = 3x + 4 + noise")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)


x = np.linspace(1, 10, 100)
y = x*x + 2*x + 1+ np.random.normal(0, 8, 100)

plt.plot(x, y)
plt.title("Non-Linear Data with random noise")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```
